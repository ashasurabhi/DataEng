```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

np.random.seed(42)
n = 50000

session_ids = [f"sess_{i}" for i in np.random.randint(1, 8000, n)]
user_ids = [f"user_{i}" if random.random() > 0.3 else None for i in np.random.randint(1, 3000, n)]  # 30% anonymous
pages = ["home", "category", "product", "cart", "checkout", "confirmation", "search", None]
events = ["page_view", "add_to_cart", "remove_from_cart", "purchase", "abandon", "search", "page_view", "page_view"]
products = [f"prod_{i}" if random.random() > 0.4 else None for i in np.random.randint(1, 500, n)]

base_time = datetime(2024, 1, 1)
timestamps = [base_time + timedelta(seconds=random.randint(0, 86400*90)) for _ in range(n)]

df = pd.DataFrame({
    "event_id":       range(n),
    "session_id":     session_ids,
    "user_id":        user_ids,
    "event_type":     np.random.choice(events, n),
    "page":           np.random.choice(pages, n),
    "product_id":     products,
    "event_time":     timestamps,
    "time_on_page_s": np.random.choice([None, -5, 0, 10, 30, 120, 300, 9999], n),  # negatives, nulls, outliers
    "device":         np.random.choice(["mobile", "desktop", "tablet", "MOBILE", "Desktop", None], n),
    "referrer":       np.random.choice(["google", "direct", "email", "social", None, "GOOGLE"], n),
    "country":        np.random.choice(["AU", "US", "UK", "au", None, "UNKNOWN"], n),
})

# Inject duplicate event_ids
df = pd.concat([df, df.sample(500)], ignore_index=True)

df.to_parquet(
    "clickstream_raw.parquet",
    index=False,
    engine="pyarrow"
)
print(f"Generated {len(df)} rows")
```
