```python
import csv
import os

# ── Create output folder ─────────────────────────────────────────────────────
output_path = "C:/spark_output/data"
os.makedirs(output_path, exist_ok=True)

# ── 1. Generate dim_employee (initial dimension table) ───────────────────────
employees = [
    (1, "E001", "John",  "Engineering", 85000, True,  "2022-01-01", ""),
    (2, "E002", "Sarah", "Marketing",   72000, True,  "2022-03-01", ""),
    (3, "E003", "Mike",  "Sales",       65000, True,  "2022-06-01", ""),
]

employee_headers = ["employee_sk", "emp_id", "name", "department", "salary", "is_current", "start_date", "end_date"]

with open(f"{output_path}/dim_employee.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(employee_headers)
    writer.writerows(employees)

print(f"dim_employee.csv created at {output_path}")

# ── 2. Generate df_updates (incoming changes) ────────────────────────────────
updates = [
    ("E001", "John",  "Engineering", 95000),  # salary increase
    ("E002", "Sarah", "Product",     78000),  # department + salary change
    ("E004", "Emma",  "HR",          60000),  # brand new employee
]

updates_headers = ["emp_id", "name", "department", "salary"]

with open(f"{output_path}/employee_updates.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(updates_headers)
    writer.writerows(updates)

print(f"employee_updates.csv created at {output_path}")

# ── 3. Verify files ──────────────────────────────────────────────────────────
for filename in ["dim_employee.csv", "employee_updates.csv"]:
    filepath = f"{output_path}/{filename}"
    with open(filepath, "r") as f:
        lines = f.readlines()
    print(f"\n{filename} ({len(lines)-1} rows):")
    for line in lines:
        print(" ", line.strip())
```
